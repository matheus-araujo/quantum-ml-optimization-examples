## 3.2.1 From classical variables to qubits
So far, the formulations that we have considered for the Max-Cut problem and for the Ising
model are purely classical. They do not mention quantum elements such as qubits, quantum
gates, or measurements. But we are closer than you might think to being able to give a
quantum formulation for these problems. We will start with a very simple instance of the
Max-Cut problem and show how we can easily transform it into quantum form. Consider
the graph in Figure 3.5. We already know that the corresponding Max-Cut problem can be
written as follows:
$$
Minimize z_0 z_1\\
Subject \ to\  z_j e \{-1, 1\} , j=0, ..., 2
$$

drawn a graph 

$\bra 0Z\ket 0 = (1 \ 0) \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}\begin{pmatrix}1 \\ 0\end{pmatrix} = 1$

$\bra 1Z\ket 1 = (0 \ 1) \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}\begin{pmatrix}0 \\ 1\end{pmatrix} = -1$

Now, consider the tensor product $𝑍 \otimes 𝑍 \otimes 𝐼$ and basis state $\ket{010}$

$$
\bra{010} ZZI\ket{010} = \bra{010} Z\ket0\otimes Z \ket1\otimes I \ket0 \\
\bra0 Z\ket0\otimes \bra1Z \ket1\otimes \bra0I \ket0 = 1(-1)1=-1

$$
